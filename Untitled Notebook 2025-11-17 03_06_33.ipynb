```python
import sys
import os
import joblib
from concurrent.futures import ThreadPoolExecutor
import time
import json
import pandas as pd
import mlflow

# --- [Configuration] ---

# 1. Ensure your code path is added
# (consistent with the path used in your train.py)
sys.path.append("/Workspace/9900-f18a-cake/mt-method2/src")

# 2. Define the path for your "Child" notebook
WORKER_NOTEBOOK_PATH = "./(Clone) child4"

# 3. Define your MLflow experiment
MLFLOW_EXPERIMENT_PATH = "/Workspace/9900-f18a-cake/classifier"

# 4. Define the path for your DiseaseTree file
# (this is obtained from your previous Parent notebook)
JOBLIB_PATH = "/Workspace/9900-f18a-cake/working_branch/data/freeze0525/diseaseTree_mapped.joblib"

# --- [End of Configuration] ---

print("Environment setup complete.")
print(f"Worker Notebook: {WORKER_NOTEBOOK_PATH}")
print(f"MLflow Experiment: {MLFLOW_EXPERIMENT_PATH}")
```

```python
# Load DiseaseTree
try:
    tree_object = joblib.load(JOBLIB_PATH)
    print("DiseaseTree node list loaded successfully.")
except Exception as e:
    print(f"Failed to load DiseaseTree: {e}")
    tree_object = None  # Ensure tree_object is defined

# Extract node names from the Tree
def get_nodes_to_train(tree):
    all_nodes = []
    if tree is None:
        return []

    def traverse(node):
        if node.name != 'ZERO2' and hasattr(node, 'samples') and len(node.samples) > 0:
             all_nodes.append(node.name)
        if hasattr(node, 'children'):
            for child in node.children:
                traverse(child)
    
    traverse(tree)
    unique_nodes = list(set(all_nodes))
    print(f"Extracted {len(unique_nodes)} unique nodes from DiseaseTree.")
    return unique_nodes

nodes_to_train = get_nodes_to_train(tree_object)
print(f"Node list (first 10): {nodes_to_train[:10]}")
```

```python
# --- [Configure Your Run] ---

# 1. Set parallelism
MAX_PARALLEL_RUNS = 10

# 2. (Testing) Select nodes to run
#    Uncomment one of the lines below for testing

# nodes_to_run = nodes_to_train[:5] # (Run the first 5)
# nodes_to_run = ["Haematological malignancy"] # (Only run the one that succeeded in your child4)
# nodes_to_run = nodes_to_train[5:10] # (Run from the 5th to the 10th)

# 3. (Production)
nodes_to_run = nodes_to_train

# --------------------------

# Set up MLflow experiment
mlflow.set_experiment(MLFLOW_EXPERIMENT_PATH)

# Start a "Parent" MLflow run to manage all child runs
with mlflow.start_run(run_name="Main_Parallel_Runner") as parent_run:
    
    parent_run_id = parent_run.info.run_id
    print(f"--- Started Main Parent Run: {parent_run_id} ---")
    print(f"--- Starting parallel training for {len(nodes_to_run)} nodes, max parallelism: {MAX_PARALLEL_RUNS} ---")

    all_results = [] # (node_name, status, json_output)
    start_time = time.time()

    # Define the task to be executed for each thread
    def train_node_parallel(node_name):
        print(f"➡️  [START] Node: {node_name}")
        
        # These parameters will be passed as Widgets to child4
        params = {
          "NODE_ID": node_name,
          "PARENT_RUN_ID": parent_run_id # Pass Parent Run ID
        }
        
        timeout_seconds = 7200 # 2 hours timeout
        
        try:
            # 3. Execute Child Notebook
            result_json = dbutils.notebook.run(WORKER_NOTEBOOK_PATH, timeout_seconds, params)
            print(f"✅  [SUCCESS] Node: {node_name}.")
            return (node_name, "Success", result_json)
        
        except Exception as e:
            # 4. Capture failed jobs
            print(f"❌  [FAILED] Node: {node_name}. Error: {e}")
            return (node_name, "Failed", str(e))

    # Use ThreadPoolExecutor to manage parallel runs
    with ThreadPoolExecutor(max_workers=MAX_PARALLEL_RUNS) as executor:
        futures = [executor.submit(train_node_parallel, node) for node in nodes_to_run]
        for future in futures:
            all_results.append(future.result())

    end_time = time.time()
    print("\n--- [ALL JOBS COMPLETE] ---")
    print(f"Total time taken: {end_time - start_time:.2f} seconds ( {(end_time - start_time)/60:.2f} minutes )")
```

```python
# --- [SUMMARY REPORT] ---
print("\n--- Summary Report ---")
success_count = 0
failed_nodes = []
all_stats_dfs = []  # Store all successful metrics

for (node_name, status, result_data) in all_results:
    if status == "Success":
        try:
            # 1. Load the returned JSON string as a Python dictionary
            result_dict = json.loads(result_data)
            
            # 2. Extract 'metrics' from the dictionary
            metrics = result_dict.get('metrics', {})

            if metrics:
                # 3. Create a single-row DataFrame from the 'metrics' dictionary
                node_stats_df = pd.DataFrame([metrics]) 
                node_stats_df['node'] = node_name  # Add node name
                node_stats_df['status'] = 'Success'
                all_stats_dfs.append(node_stats_df)
            else:
                # The task succeeded but was skipped (e.g., insufficient samples)
                print(f"Info: Node {node_name} finished but returned no metrics (Status: SKIPPED).")
                df = pd.DataFrame([{"node": node_name, "status": "SKIPPED"}])
                all_stats_dfs.append(df)
            
            success_count += 1

        except Exception as e:
            print(f"Warning: Could not parse results from node {node_name}: {e}")
            failed_nodes.append((node_name, f"Parse Error: {e}"))
            
    else:
        # The task failed at the dbutils.notebook.run level
        failed_nodes.append((node_name, result_data))

print(f"\nSuccessful Jobs: {success_count} / {len(nodes_to_run)}")
print(f"Failed Jobs: {len(failed_nodes)} / {len(nodes_to_run)}")
if failed_nodes:
    print("\n--- List of Failed Jobs ---")
    for (node, error_msg) in failed_nodes:
        print(f"Node: {node}\nError: {error_msg}\n")

# 4. Merge all successful results and display them
if all_stats_dfs:
    final_stats_summary = pd.concat(all_stats_dfs, ignore_index=True)
    print("\n--- [All Model Performance Metrics] ---")
    display(final_stats_summary)
else:
    print("\n--- [All Model Performance Metrics] ---")
    print("No performance metrics were successfully collected.")
```
