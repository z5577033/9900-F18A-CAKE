```python
import unittest
import sys

modules_to_reload = [
    'mch.tests.test_config',
    'mch.tests.test_database',
    'mch.tests.test_model_generation'
]
for mod in modules_to_reload:
    if mod in sys.modules:
        del sys.modules[mod]

try:
    loader = unittest.TestLoader()
    suite = unittest.TestSuite()
    
    from mch.tests import test_database
    suite.addTests(loader.loadTestsFromModule(test_database))
    
    from mch.tests import test_model_generation
    suite.addTests(loader.loadTestsFromModule(test_model_generation))
    
    print("🚀 Starting Full Test Suite...\n")
    unittest.TextTestRunner(verbosity=2).run(suite)
    
except Exception as e:
    print(f"❌ Error: {e}")
```
