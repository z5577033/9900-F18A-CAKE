```python
import sys
import os
import joblib

# --- [CONFIGURATION] ---
# Path to your source code
sys.path.append("/Workspace/9900-f18a-cake/mt-method2/src")

# Path to the DiseaseTree joblib file
JOBLIB_PATH = "/Workspace/9900-f18a-cake/mt-method2/data/freeze0525/diseaseTree_mapped.joblib"
# -----------------------

print("Loading DiseaseTree...")

try:
    tree_object = joblib.load(JOBLIB_PATH)
    print("✅ DiseaseTree loaded successfully.")
except Exception as e:
    print(f"❌ Failed to load DiseaseTree: {e}")
    tree_object = None

def get_nodes_to_train(tree):
    """
    Traverses the tree to find all applicable nodes for training.
    """
    all_nodes = []
    if tree is None: return []
    
    def traverse(node):
        # Logic: Keep non-root nodes that have samples
        if node.name != 'ZERO2' and hasattr(node, 'samples') and len(node.samples) > 0:
             all_nodes.append(node.name)
        # Recurse to children
        if hasattr(node, 'children'):
            for child in node.children:
                traverse(child)
    
    traverse(tree)
    return list(set(all_nodes))

# Get the full list of nodes
nodes_to_train = get_nodes_to_train(tree_object)

# --- [TESTING VS PRODUCTION] ---
# Uncomment the line below for testing (e.g., run only the first 3 nodes)
nodes_to_run = nodes_to_train[:3] 

# Uncomment the line below for FULL PRODUCTION (run all 217 nodes)
# nodes_to_run = nodes_to_train 

print(f"--- Ready to submit {len(nodes_to_run)} jobs ---")
```

```python
import time
from databricks.sdk import WorkspaceClient
from databricks.sdk.service.jobs import SubmitTask, NotebookTask

# 1. Initialize SDK (Standard mode, uses the notebook's internal auth)
w = WorkspaceClient()

# 2. Define path
CHILD_NOTEBOOK_PATH = "/Workspace/9900-f18a-cake/mt-method2/(Clone) child4"

# 3. Get the CURRENT Cluster ID
#    Since we cannot create new clusters (no token permissions),
#    we will run the jobs on THIS cluster.
try:
    CURRENT_CLUSTER_ID = spark.conf.get("spark.databricks.clusterUsageTags.clusterId")
    print(f"ℹ️ Using Existing Cluster ID: {CURRENT_CLUSTER_ID}")
except Exception as e:
    print("❌ Could not detect current cluster ID. Please hardcode it.")
    raise e

print(f"--- Submitting {len(nodes_to_run)} parallel jobs via Databricks SDK ---")
print("⚠️ Note: Running on EXISTING cluster due to permission limits.")

submitted_jobs = []

for node in nodes_to_run:
    safe_node_name = node.replace(' ', '_').replace(',', '')
    run_name = f"Train_{safe_node_name}"
    
    print(f"➡️ Submitting Job for node: {node} ...")
    
    try:
        # 4. Submit the Run using existing_cluster_id
        run = w.jobs.submit(
            run_name=run_name,
            tasks=[
                SubmitTask(
                    task_key="training_task",
                    existing_cluster_id=CURRENT_CLUSTER_ID, # <--- CRITICAL CHANGE
                    notebook_task=NotebookTask(
                        notebook_path=CHILD_NOTEBOOK_PATH,
                        base_parameters={
                            "NODE_ID": node
                        }
                    )
                )
            ]
        )
        
        print(f"   ✅ Success! Run ID: {run.run_id}")
        print(f"   🔗 Link: {run.run_page_url}")
        submitted_jobs.append((node, run.run_id, run.run_page_url))
        
        # Sleep briefly
        time.sleep(0.5)
        
    except Exception as e:
        print(f"   ❌ Failed to submit job for {node}: {e}")

print(f"\n--- [COMPLETE] Submitted {len(submitted_jobs)} / {len(nodes_to_run)} jobs ---")
print("👉 Please go to the 'Workflows' -> 'Job runs' tab to see them running.")
```
